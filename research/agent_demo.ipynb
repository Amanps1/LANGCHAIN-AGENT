```python
import os
import certifi
from dotenv import load_dotenv

from langchain.tools import tool
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_groq import ChatGroq
from langchain import hub
from langchain.tools import tool
import requests
```

```python
from langchain.agents import create_react_agent, AgentExecutor
```

```python
load_dotenv()
```

```python
os.environ["SSL_CERT_FILE"] = certifi.where()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
WEATHERSTACK_API_KEY = os.getenv("WEATHERSTACK_API_KEY")
os.environ["GROQ_API_KEY"] = GROQ_API_KEY
os.environ["TAVILY_API_KEY"] = TAVILY_API_KEY
os.environ["WEATHERSTACK_API_KEY"] = WEATHERSTACK_API_KEY
```

```python
search_tool = TavilySearchResults(
    max_results=2
)
```

```python
@tool
def get_weather_data(city: str) -> str:
    """
    Get current weather information for a specified city.

    Args:
        city: Name of the city to get weather details.

    Returns:
        Temperature, weather condition, and humidity information.
    """

    url = (
        "https://api.weatherstack.com/current?"
        f"access_key={WEATHERSTACK_API_KEY}"
        f"&query={city}"
    )

    try:
        response = requests.get(url)
        data = response.json()
        if "current" not in data:
            return f"Could not fetch weather data for {city}. Error: {data}"

        current = data["current"]

        return (
            f"Weather Report for {city}\n"
            f"Temperature: {current['temperature']}°C\n"
            f"Condition: {current['weather_descriptions'][0]}\n"
            f"Humidity: {current['humidity']}%"
        )

    except Exception as e:
        return f"Error occurred: {str(e)}"
```

```python
search_tool.invoke("Give me the latest news on AI?")
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0,
    api_key=GROQ_API_KEY
)
```

```python
response=llm.invoke("Give me the latest news on AI?")
```

```python
response
```

```python
prompt = hub.pull(
    "hwchase17/react"
)
```

```python
tools=[search_tool, get_weather_data]
```

```python
agent=create_react_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
```

```python
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)
```

```python
response=agent_executor.invoke({
    "input":"Find the capital of france and then get the current weather in that city."
})
```
